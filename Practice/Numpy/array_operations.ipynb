Stacking Array

```python
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
```

```python
np.vstack((a,b))
# One array stacked below another array
```

```python
np.hstack((a,b))
# One array stacked horizontally/along with the previous array like->string concatination'+'
```

```python
np.column_stack((a,b))
# Stacking arrays along columns
```

Splitting Array

```python
c=np.arange(16).reshape(4,4)
c
```

```python
np.split(c,2)
# Splits an array into equal-sized subarrays.
# If the split is not possible evenly, it raises an error.
```

```python
np.array_split(c,3)
# This Similar to split(), but allows uneven splits without error.
```

```python
# Split rows into 2 equal parts
rows = np.split(c, 2, axis=0)

# Split columns into 2 equal parts
cols = np.split(c, 2, axis=1)

print("Row split:", rows)
print("Column split:", cols)
```

```python
np.vsplit(c,4)
```

```python
np.hsplit(c,2)
```
