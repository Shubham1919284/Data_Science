```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib  inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
iris
```

```python
iris.data
```

```python
iris.target
```

```python
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
df
```

```python
#independent feature and dependent features
X=df.iloc[:,:-1]
y=iris.target
```

```python
X,y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.33,random_state=42)
```

```python
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
#Post-Prunning
treemodal=DecisionTreeClassifier()
```

```python
treemodal.fit(X_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treemodal,filled=True)
plt.show()
```

```python
treemodal1=DecisionTreeClassifier(max_depth=2)
treemodal1.fit(X_train,y_train)
```

```python
plt.figure(figsize=(15,10))
tree.plot_tree(treemodal1,filled=True)
plt.show()
```

```python
y_pred=treemodal.predict(X_test)
y1_pred=treemodal1.predict(X_test)
y_pred
```

```python
y1_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
score=accuracy_score(y_pred,y_test)
print(score)
```

```python
score1=accuracy_score(y1_pred,y_test)
print(score1)
```

```python
print(classification_report(y_pred,y_test))
```

```python
print(classification_report(y1_pred,y_test))
```
